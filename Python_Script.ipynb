```python
#Containers - List, Dictionaries, Tuples, Sets: to store multiple values (similar to array) collection of data
list1=[3,12,11,14,9]
print(list1, list1[4], list1[-1]) #index starts with 0, can go right(positive) and go left(negative)

print(len(list1))
print(list1[2]) #fetching data from index 2
```

```python
list1[3]='data'  #replacing the value present at index 3
print(list1) #so you can put multiple data types inside a list
```

```python
list1.append('Science')  #always added after/behind the list
print(list1)
#The size of list is flexible, can add and remove elements in a list
list1.pop(2) #deletes the last element if left blank
print(list1)
```

```python
numbers=list(range(6)) #not including 6
print(numbers)
```

```python
#slicing
list2=[11,2,5,18,3,20]    #index: element
print(list2[2:4]) #not print the last element, print value at index 2, 3
print(list2[2:]) #including 2, after 2 everything
print(list2[:4])
print(list2[:])
print(list2[:-2])
list2[2:4]= [50,100]
print(list2)
```

```python
#dictionary
d={'cat':'cute', 'dog':'furry'} #key: value, key can be of any type, using curly braces
print(d['cat']) #printing the value of 'cat' key
print('cute' in d) #check whether the key is inside the dictionary
```
