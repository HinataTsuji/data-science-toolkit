```python
import numpy as np  #library to deal with multidimensional arrays
import random

#Create a numpy array of size 2 x 2 (2 rows and 2 columns)
a = np.array([[1,2],[3,4]])
print(a)
```

```python
#Create a random numpy array of size 3 x 4
b = np.random.randint(1, 101, size=(3,4))
b2 = np.random.rand(3,4) #range of random is from 0 to 1
print(b)
print(f'\n{b2}')
```

```python
#Create a numpy array of ones
c = np.ones((3,3))
#Create a numpy array of zeroes
d = np.zeros((3,3))
print(c)
print(d)
```

```python
# Create a random numpy array of size 6x3 and transpose it
e = np.random.rand(6,3)
print(e)

f = np.transpose(e) #invert an array rows become columns, columns become rows
f2 = e.T  #transpose
print(f)
print('\n',f2)
```

```python
# Create two random numpy array of size n x p and p x m
# and then compute the dot product

# Define the dimensions
# In this example, it is A = 2 x 3 matrix, B = 3 x 4 matrix
n = 2 # Number of rows for the first array
p = 3 # Number of columns for the first array and rows for the second array
m = 4 # Number of columns for the second array

# Create the random arrays and calculate the dot product
g = np.random.random((n,p))
h = np.random.random((p,m))
dotprod = np.dot(g, h)
print(g)
print(h)
print(dotprod)
```
