```python
df.hist()
```

```python
# can remove using stats function from scipy library
from scipy import stats
import numpy as np
df[(np.abs(stats.zscore(df)) < 3).all(axis=1)] # calculates z score for every value in dataframe
```

Here is a breakdown of the entire expression:
stats.zscore(df): Calculates the Z-score for every value in the DataFrame, relative to its own column's mean and standard deviation.
np.abs(...): Takes the absolute value of the Z-scores, as the direction (above or below the mean) does not matter, only the magnitude of deviation.
< 3: Creates a new DataFrame of boolean values (True/False) where True indicates that the absolute Z-score for a specific cell is less than 3 (i.e., within 3 standard deviations of the mean), and False indicates it is an outlier.
.all(axis=1): This is the key part.
axis=1 refers to the columns of the DataFrame (operating along the horizontal axis for each row).
.all() checks if all values in a given row are True.
Therefore, it returns a single True for a row only if every single column in that row satisfies the condition (absolute Z-score < 3). If even one column has an outlier (absolute Z-score >= 3), the result for that row is False.
df[...]: Uses the resulting boolean Series to filter the original DataFrame, keeping only the rows where .all(axis=1) was True, effectively removing any row that had at least one outlier in any column
